```python
import random

def calculate_conflicts(board, n, k):
    conflicts_num = 0
    queens_position_arr = []

    for i in range(n):
        for j in range(n):
            if board[i][j] == 'Q':
                queens_position_arr.append((i, j))

    for i in range(k):
        row1, col1 = queens_position_arr[i]

        for j in range(i+1, k):
            row2, col2 = queens_position_arr[j]

            if col1 == col2:
                conflicts_num += 1

            if abs(row1-row2) == abs(col1-col2):
                conflicts_num += 1

    return conflicts_num


def get_neighbors(board, n, k):

    neighbors_arr = []

    for i in range(n):
        for j in range(n):

            if board[i][j] == 'Q':

                for updated_col in range(n):

                    if updated_col != j:

                        new_board = []

                        for a in range(n):
                            new_row = []
                            for b in range(n):
                                new_row.append(board[a][b])
                            new_board.append(new_row)

                        new_board[i][j] = '.'
                        new_board[i][updated_col] = 'Q'

                        neighbors_arr.append(new_board)

    return neighbors_arr


def hill_climbing_n_queens(board, n, k):

    current_position = board
    current_conflicts = calculate_conflicts(current_position, n, k)

    while True:

        neighbors_var = get_neighbors(current_position, n, k)

        best = current_position
        best_conflicts = current_conflicts

        for i in neighbors_var:

            conf = calculate_conflicts(i, n, k)

            if conf < best_conflicts:
                best = i
                best_conflicts = conf

        if best_conflicts >= current_conflicts:
            break

        current_position = best
        current_conflicts = best_conflicts

        if current_conflicts == 0:
            break

    return current_position, current_conflicts


def solve_n_queens(n, k, initial_board):

    print("Initial Board:")
    print("[")

    for i in initial_board:
        print("  [", end=" ")
        for j in i:
            print(j, end=" ")
        print("]")

    print("]")

    conf = calculate_conflicts(initial_board, n, k)
    print("Initial Conflicts:", conf)

    final_board, final_conflicts = hill_climbing_n_queens(initial_board, n, k)

    print("Final Board:")
    print("[")

    for i in final_board:
        print("  [", end=" ")
        for j in i:
            print(j, end=" ")
        print("]")

    print("]")

    if final_conflicts == 0:
        print("Solution Found!")
    else:
        print("Local optimum reached")

    print("Final Conflicts:", final_conflicts)


n = 6
k = 6

initial_board = [
['.', 'Q', '.', '.', '.', '.'],
['.', 'Q', '.', '.', '.', '.'],
['.', '.', '.', 'Q', '.', '.'],
['.', '.', '.', '.', '.', 'Q'],
['.', '.', '.', 'Q', '.', '.'],
['.', '.', '.', '.', '.', 'Q']
]

solve_n_queens(n, k, initial_board)
```
