# Control Flow
## Conditional statements: if, elif, else
The Python syntax for conditional execution of code uses the keywords `if`, `elif` (else if), `else`:

```python
x = 10

if x > 5:
    print("x is greater than 5")
```

```python
x = 3

if x > 5:
    print("x is greater than 5")
else:
    print("x is less than or equal to 5")
```

```python
x = 3

if x > 5:
    print("x is greater than 5")
elif x == 5:
    print("x is equal to 5")
else:
    print("x is less than 5")
```

# Loops
In Python, loops can be programmed in a number of different ways. The most common is the `for` loop, which is used together with iterable objects, such as lists. The basic syntax is:
## for loops:


```python
#with lists
fruits = ['apple', 'banana', 'cherry']

for fruit in fruits:
    print(fruit)
```

```python
#with strings
word = "Python"

for letter in word:
    print(letter)
```

```python
#with range
for x in range(4): # by default range start at 0
    print (x) 
```

```python
for x in range(-3,3):
    print (x)
```

```python
#To iterate over key-value pairs of a dictionary:
params = {'name': 'John', 'age': 25, 'city': 'New York'}
for key, value in params.items():
    print (key + " = " + str(value))
```

```python
#You can use enumerate() to iterate over both the values and their indices.
fruits = ['apple', 'banana', 'cherry']

for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")
```

```python
#creating a list using for loop
l1 = [x**2 for x in range(0,5)]

print (l1)
```

## while loops

```python
i = 0

while i < 5:
    print(i)
    
    i = i + 1
    
print ("done")
```

# Functions
A function in Python is defined using the keyword `def`, followed by a function name, a signature within parentheses `()`, and a colon `:`. The following code, with one additional level of indentation, is the function body.

```python
def func0():   
    print ("test")
```

```python
func0()
```

Optionally, but highly recommended, we can define a so called "docstring", which is a description of the functions purpose and behaivor. The docstring should follow directly after the function definition, before the code in the function body.

```python
def func1(s):
    """
    Print a string 's' and tell how many characters it has    
    """
    
    print (s + " has " + str(len(s)) + " characters")
```

```python
func1("hello world")
```

Functions that returns a value use the `return` keyword:

```python
def square(x):
    """
    Return the square of x.
    """
    return x ** 2
```

```python
square(5)
```

We can return multiple values from a function using tuples (see above):

```python
def powers(x):
    """
    Return a few powers of x
    """
    return x ** 2, x ** 3, x ** 4
```

```python
powers(3)
```

### Default argument and keyword arguments
You can set *default values* for function parameters. If a value is not provided when calling the function, the default value will be used.


```python
def power(base, exponent=2):
    result = base ** exponent
    return result
```

```python
power(4)
```

```python
power(3, 3)
```

You can pass arguments to a function by explicitly naming the parameter and providing the value. This allows you to pass values out of order or skip some arguments.

```python
def print_person_info(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)
```

```python
print_person_info(age=25, name="John", city="New York")
```

## Variable Number of Arguments:
You can use *args to allow a function to accept a variable number of positional arguments.

```python
def sum_numbers(*args):
    result = sum(args)
    return result
```

```python
sum_numbers(1, 2, 3, 4, 5)
```

## Variable Number of Keyword Arguments:
You can use **kwargs to allow a function to accept a variable number of keyword arguments.

```python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(key + ": " + str(value))
```

```python
print_info(name="Alice", age=30, city="Wonderland")
```

# Classes
Classes are the key features of object-oriented programming. A class is a structure for representing an object and the operations that can be performed on the object. 

In Python a class can contain *attributes* (variables) and *methods* (functions).

A class is defined almost like a function, but using the `class` keyword, and the class definition usually contains a number of class method definitions (a function in a class).

* Each class method should have an argument `self` as its first argument. This object is a self-reference.

* Some class method names have special meaning, for example:

    * `__init__`: The name of the method that is invoked when the object is first created.
    * `__str__` : A method that is invoked when a simple string representation of the class is needed, as for example when printed.
    * There are many more, see http://docs.python.org/2/reference/datamodel.html#special-method-names

### Defining a Class:
To define a class, you use the class keyword. Inside the class, you can have attributes (variables) and methods (functions).

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number
        self.marks = {}

    def add_marks(self, subject, marks):
        self.marks[subject] = marks
    def total_marks(self):
        total_marks =sum(self.marks.values())
        return total_marks
```

### Creating Object

You can create instances (objects) of a class by calling the class as if it were a function:

```python
student1 = Student("John", "S123")
student2 = Student("Jane", "S124")
```

### Accessing Attributes and Calling Methods:
You can access attributes and call methods using dot notation:

```python
marks_1 = student1.add_marks("Math", 85)
marks_2 = student1.add_marks("English", 90)
total_marks = student1.total_marks()
print(f"{student1.name}'s marks: {total_marks}")
```

# **Class Inheritance**
( **is a** relationship.)

*   Inheritance allows us to define a class that inherits all the methods and properties from another class.
*   Classes that inherit from another are called **derived classes**, **subclasses**, or **subtypes**.

Any class can be made into a parent class, so both (parent and child class) are each fully functional classes in their own right, rather than just templates.


```python
#Parent Class
class Fish:
  #Constructor method
    def __init__(self, first_name, last_name,
                 skeleton="bone", eyelids=False):
        self.first_name = first_name
        self.last_name = last_name
        self.skeleton = skeleton
        self.eyelids = eyelids

    def swim(self):
        print("The fish is swimming.")

    def swim_backwards(self):
        print("The fish can swim backwards.")
```

```python
#Child Class
class Trout1(Fish):
    pass

fish_firstname= input("Enter Fish First Name: ")
fish_lastname= input("Enter Fish Last Name: ")
terry1 = Trout1(fish_firstname,fish_lastname)
print(terry1.first_name + " " + terry1.last_name)
print(terry1.skeleton)
print(terry1.eyelids)
terry1.swim()
terry1.swim_backwards()
```

** The super() Function**

With the **super()** function, we can gain access to inherited methods that have been overwritten in a class object.

When we use the **super()** function, we are calling a parent method into a child method to make use of it.

For example, we may want to override one aspect of the parent method with certain functionality, but then call the rest of the original parent method to finish the method.

```python
#The Parent Class
class Fish:
    def __init__(self, first_name, last_name,
                 skeleton="bone", eyelids=False):
        self.first_name = first_name
        self.last_name = last_name
        self.skeleton = skeleton
        self.eyelids = eyelids

    def swim(self):
        print("The fish is swimming.")

    def swim_backwards(self):
        print("The fish can swim backwards.")


#The super() Function
class Trout(Fish):
    def __init__(self, water = "freshwater"):
        self.water = water
        a=super().__init__(self,fish_lastname)
        print(a)



terry = Trout()

# Initialize first name
fish_firstname= input("Enter Fish First Name: ")
fish_lastname= input("Enter Fish Last Name: ")
terry.first_name = fish_firstname

# Use parent __init__() through super()
print(terry.first_name + " " + terry.last_name)
print(terry.eyelids)

# Use child __init__() override
print(terry.water)

# Use parent swim() method
terry.swim()
```

**Creating Polymorphic Classes**

```python
class Fish:
    def __init__(self, name):
        self.name = name

    def swim(self):
        raise NotImplementedError("Subclass must implement this method")

    def swim_backwards(self):
        raise NotImplementedError("Subclass must implement this method")

    def skeleton(self):
        raise NotImplementedError("Subclass must implement this method")


class Shark(Fish):
    def swim(self):
        print(f"The {self.name} is swimming.")

    def swim_backwards(self):
        print(f"The {self.name} cannot swim backwards, but can sink backwards.")

    def skeleton(self):
        print(f"The {self.name} skeleton is made of cartilage.")


class Clownfish(Fish):
    def swim(self):
        print(f"The {self.name} is swimming.")

    def swim_backwards(self):
        print(f"The {self.name} can swim backwards.")

    def skeleton(self):
        print(f"The {self.name} skeleton is made of bone.")


def in_the_pacific(fish):
    fish.swim()


# Create instances of Shark and Clownfish
sammy = Shark("Sammy the Shark")
casey = Clownfish("Casey the Clownfish")

# Call the methods of Shark and Clownfish
sammy.swim()
sammy.swim_backwards()
sammy.skeleton()

casey.swim()
casey.swim_backwards()
casey.skeleton()

# Use polymorphism with the in_the_pacific function
in_the_pacific(sammy)
in_the_pacific(casey)
```
