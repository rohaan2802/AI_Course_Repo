# 🔬Simple Reflex Agents
<div style="background-color:#fff1cc; padding: 20px;">
<p style="font-size:20px; font-family:verdana; line-height: 1.7em">Simple reflex agents are a type of AI agent that selects actions based solely on the current percept (the agent's current view of the environment). These agents do not have memory or the ability to consider past perceptions. They operate using a set of condition-action rules, also known as "if-then" rules.

* Simple Reflex Agents are useful in fully observable environments where all necessary information is available at each step. However, they struggle in partially observable environments because they do not maintain internal memory.

# **Enviroment**
### **1. Observable**
  * The agent **can see the entire state** of the environment at each step.
### **2. Partially Observable**
  * The agent **cannot see the entire environment**; it has limited information.
### **3. Deterministic**
  * The next state is **completely** determined by the **current state and action**.
### **3. Stochastic**
  * Next state is uncertain; there is some **randomness**.









# **Basic Concept**
### **Percept**
It represents what the agent perceives about the current state of the environment.

The agent uses percepts to decide its next action.

`e.g. env.get_opponent_move() `

### **Action**
What agent does in reponse

`e.g. agent.choose_action()`

# **Example**


#### 1. **Light Bulb Example:**

#####  **Environment**: The Room and Its Brightness
-  The environment in this case is the room, and it has one property: brightness.
-  The brightness of the room can be either "Dark" or "Bright".

#####  *Percepts*: How the Agent Senses the Environment
#####  *Actions*: How the Agent Responds to the Environment

```python
class LightSwitch:
    def __init__(self, state):
        #initial state of a bulb
        self.state=state
    def percept(self):
        #current state of the environment
        return self.state
    def action(self):
        #perform the actions based on a percept

        if self.percept()=="Dark":
            print("the room is dark. turning on the light")
            self.state= "Bright"
        else:
            print("Room is bright. Turning the Lights Off")
            self.state= "Dark"
```

```python
#creating instance of the agent
room = LightSwitch("Dark")  # Room starts dark
print("Initial State:", room.state)

# Agent acts
room.action()
print("Updated State:", room.state)

# Agent acts again
room.action()
print("Updated State:", room.state)
```

#### Implement a model-based reflex agent for a simple game like Rock, Paper, Scissors.

```python
import random

class Environment:
    def __init__(self):
        self.moves = ['rock', 'paper', 'scissors']
        self.opponent_last_move = None

    def get_opponent_move(self):
        self.opponent_last_move = random.choice(self.moves)
        return self.opponent_last_move

```

```python
class ModelBasedReflexAgent:
    def __init__(self, environment):
        self.environment = environment
        self.model = {'opponent_last_move': None}

    def update_model(self):
        self.model['opponent_last_move'] = self.environment.get_opponent_move()


    def choose_action(self):

        if self.model['opponent_last_move'] == 'rock':
            return 'paper'
        elif self.model['opponent_last_move'] == 'paper':
            return 'scissors'
        else:
            return 'rock'
```

```python
environment = Environment()
agent = ModelBasedReflexAgent(environment)

# Opponent makes a move
agent.update_model()
# Agent makes its decision based on opponent's last move
action = agent.choose_action()
opponent_move = agent.model['opponent_last_move']

print(f"Opponent's move: {opponent_move}")
print(f"Agent's move: {action}")
```

```python
import random
from collections import Counter

# Environment
class Environment:
    def __init__(self):
        self.moves = ['rock', 'paper', 'scissors']

    def get_opponent_move(self):
        return random.choice(self.moves)

# Goal-Based Agent
class GoalBasedAgent:
    def __init__(self, environment):
        self.env = environment
        self.history = []  # Track opponent's moves

    def update_history(self, move):
        self.history.append(move)

    def choose_action(self):
        if not self.history:
            return random.choice(self.env.moves)
        # Predict opponent's most frequent move and counter it
        most_common = Counter(self.history).most_common(1)[0][0]
        return {'rock':'paper', 'paper':'scissors', 'scissors':'rock'}[most_common]

# Simulation
env = Environment()
agent = GoalBasedAgent(env)

agent_wins = 0
opponent_wins = 0
rounds = 10

for i in range(1, rounds+1):
    opponent = env.get_opponent_move()
    agent.update_history(opponent)
    agent_move = agent.choose_action()

    # Determine result
    if (agent_move == 'rock' and opponent == 'scissors') or \
       (agent_move == 'paper' and opponent == 'rock') or \
       (agent_move == 'scissors' and opponent == 'paper'):
        agent_wins += 1
        result = "Agent Wins"
    elif agent_move == opponent:
        result = "Draw"
    else:
        opponent_wins += 1
        result = "Opponent Wins"

    print(f"Round {i}: Opponent: {opponent}, Agent: {agent_move}, Result: {result}")

# Announce overall winner
print("\n--- Final Result ---")
print(f"Agent Wins: {agent_wins}, Opponent Wins: {opponent_wins}")
if agent_wins > opponent_wins:
    print("Overall Winner: Agent :) ")
elif agent_wins < opponent_wins:
    print("Overall Winner: Opponent :)")
else:
    print("Overall Result: Draw ")
```
