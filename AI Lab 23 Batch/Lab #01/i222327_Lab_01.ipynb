# Q# 02 Write a Python function that takes a list of numbers as input and returns the count of occurrences for each unique number in the list.

```python
def count_fun(num_list):
    counts = {}

    for i in num_list:
        if i in counts:
            counts[i] += 1
        else:
            counts[i] = 1

    return counts


num_list = [1, 2, 3, 2, 1, 4, 3]
result = count_fun(num_list)

for j, val in result.items():
    print(f"{j}: {val}")
```

# Q# 03. Write a function that compares your and your friend's information, including name, age,and list of subjects. The function should:
# • Identify the subjects that you are studying but your friend is not.
# • Determine whether you are older than your friend.

```python
def compare_info(your_info, friend_info):

    your_subjects = set(your_info["subjects"])
    friend_subjects = set(friend_info["subjects"])

    extra_subjects = your_subjects - friend_subjects

    print("Subjects you are studying but your friend is not:")
    for subject in extra_subjects:
        print(subject)


    if your_info["age"] > friend_info["age"]:
        print("You are elder than your friend.")
    elif your_info["age"] < friend_info["age"]:
        print("Your friend is elder than you.")
    else:
        print("Both are of the same age.")


your_info = {
    "name": "xyz",
    "age": 23,
    "subjects": ["oop", "dsa", "pf", "os", "automata", "compiler", "web"]
}

friend_info = {
    "name": "abc",
    "age": 20,
    "subjects": ["oop", "dsa", "compiler", "web"]
}


compare_info(your_info, friend_info)
```

# Q# 04. Write a program that squares each value in a given list, then checks if the squared valueis even or odd. Print the result in a readable format.

```python
numbers = [1, 2, 3, 4]

for num in numbers:
    square = num * num
    if square % 2 == 0:
        print(f"{square}: even")
    else:
        print(f"{square}: odd")
```

# Q# 05. Create a base class Vehicle and derived classes Car and Bike.
# • Create a base class Vehicle with:
# ▪ Attributes : make (manufacturer), model, and year.
# ▪ Method : display_info() that prints the details of the vehicle.

# • Create a derived class Car that inherits from Vehicle:
# ▪ Attributes : Add an attribute num_doors (number of doors).
# ▪ Method : Override the display_info() method to include num_doors. •
# Create another derived class Bike that inherits from Vehicle:
# ▪ Attributes : Add an attribute type_of_bike (e.g., "mountain", "road"). ▪
# Method : Override the display_info() method to include type_of_bike.
# Create objects for both Car and Bike, set their attributes, and call the display_info() method to display the information.

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print("Make:", self.make)
        print("Model:", self.model)
        print("Year:", self.year)


class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors

    def display_info(self):
        print("Car Details:")
        super().display_info()
        print("Number of Doors:", self.num_doors)
        print()


class Bike(Vehicle):
    def __init__(self, make, model, year, type_of_bike):
        super().__init__(make, model, year)
        self.type_of_bike = type_of_bike

    def display_info(self):
        print("Bike Details:")
        super().display_info()
        print("Type of Bike:", self.type_of_bike)
        print()


car = Car("Toyota", "Corolla", 2020, 4)
bike = Bike("Trek", "Marlin 7", 2021, "Mountain")

car.display_info()
bike.display_info()
```

# Q# 06. Create a Python class for a rectangle that can calculate and return both the area and diagonal of the rectangle.

# Area = length ∗ width

# • Create an object of the Rectangle class with specific length and width • Call the area()and diagonal() methods, and print the results.

```python
import math

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def diagonal(self):
        return math.sqrt(self.length**2 + self.width**2)


rect = Rectangle(6, 8)

print("Area of Rectangle:", rect.area())
print("Diagonal of Rectangle:", rect.diagonal())
```

# Q# 07. Write a Python program for a library management system using class inheritance. Design a base class ‘LibraryItem’ with attributes like ‘item_id’,‘title’, and ‘availability’. Implement methods, including ‘check_out()’ and ‘return_item()’, to manage borrowing and returning items. Derive two classes, ‘Book’ and ‘ResearchThesis’, inheriting from ‘LibraryItem’. Include specific attributes (e.g.,‘author’ for books, ‘Thesis_Topic’ for ResearchThesis) and customize the ‘get_info()’ method in each derived class for additional information. Implement methods such as ‘calculate_penalty()’ for late returns. To simulate library operations, create a function, ‘simulate_library_operations(library_items)’, that takes a list of item instances (mix of books and ResearchThesis) and demonstrates operations like checking out and returning items, displaying information, and calculating penalties for late returns. Note: Ensure to focus on the class inheritance, method overriding, and polymorphic behavior while designing this library management system.

```python

class LibraryItem:
    def __init__(self, item_id, title, availability=True):
        self.item_id = item_id
        self.title = title
        self.availability = availability

    def check_out(self):
        if self.availability:
            self.availability = False
            print(f"Checked out: {self.title}")
        else:
            print(f"'{self.title}' is already checked out.")

    def return_item(self):
        if not self.availability:
            self.availability = True
            print(f"Returned: {self.title}")
        else:
            print(f"'{self.title}' is already available.")

    def get_info(self):
        status = "Available" if self.availability else "Checked Out"
        return f"ID: {self.item_id}\nTitle: {self.title}\nStatus: {status}"

    def calculate_penalty(self, days_late):
        if days_late <= 0:
            return 0
        return days_late * 10


class Book(LibraryItem):
    def __init__(self, item_id, title, author, availability=True):
        super().__init__(item_id, title, availability)
        self.author = author

    def get_info(self):
        base = super().get_info()
        return base + f"\nAuthor: {self.author}\nType: Book"

    def calculate_penalty(self, days_late):
        if days_late <= 0:
            return 0
        return days_late * 5

class ResearchThesis(LibraryItem):
    def __init__(self, item_id, title, thesis_topic, availability=True):
        super().__init__(item_id, title, availability)
        self.thesis_topic = thesis_topic

    def get_info(self):
        base = super().get_info()
        return base + f"\nThesis Topic: {self.thesis_topic}\nType: ResearchThesis"

    def calculate_penalty(self, days_late):
        if days_late <= 0:
            return 0
        return days_late * 20


def simulate_library_operations(library_items):
    print("===== LIBRARY ITEMS INFO (Polymorphism: get_info) =====")
    for item in library_items:
        print(item.get_info())
        print("-" * 40)

    print("\n===== CHECKING OUT ITEMS =====")
    for item in library_items:
        item.check_out()

    print("\n===== TRY CHECKING OUT AGAIN (should show already checked out) =====")
    for item in library_items:
        item.check_out()

    print("\n===== RETURNING ITEMS =====")
    for item in library_items:
        item.return_item()

    print("\n===== PENALTY CALCULATION (Polymorphism: calculate_penalty) =====")
    days_late_list = [0, 2, 5]
    for item, days_late in zip(library_items, days_late_list):
        penalty = item.calculate_penalty(days_late)
        print(f"{item.title} | Days late: {days_late} | Penalty: {penalty}")


book1 = Book(101, "OOP in Python", "Ali Khan")
book2 = Book(102, "Data Structures", "Sara Ahmed")
thesis1 = ResearchThesis(201, "AI Research", "Deep Learning in Healthcare")

library_items = [book1, book2, thesis1]

simulate_library_operations(library_items)
```
