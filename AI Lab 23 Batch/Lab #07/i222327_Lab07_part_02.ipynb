```python
import random

# Item characteristics - weights and values
weight_list = [15, 40, 15, 20]
value_list = [30, 100, 40, 50]
capacity = 50
num_items = 4

population_size = 6
generations = 10
random.seed(42)

def generate_chromosome():
    # Create a random solution (chromosome)
    chromosome = []
    for i in range(num_items):
        chromosome.append(random.randint(0, 1))  # 0 means don't take item, 1 means take it
    return chromosome


def calculate_fitness(chromosome):
    total_weight = 0
    total_value = 0

    # Loop through each item
    for i in range(num_items):
        if chromosome[i] == 1:
            total_weight += weight_list[i]
            total_value += value_list[i]

    # Penalty for exceeding capacity - just return 0 fitness
    if total_weight > capacity:
        return 0

    return total_value


def selection(population):
    # Tournament selection - pick 2 random chromosomes and return the better one
    c1 = random.choice(population)
    c2 = random.choice(population)
    fitness1 = calculate_fitness(c1)
    fitness2 = calculate_fitness(c2)

    if fitness1 > fitness2:
        return c1
    else:
        return c2


def crossover(parent1, parent2):
    # Single point crossover
    split_point = random.randint(1, num_items - 1)

    child1 = parent1[:split_point] + parent2[split_point:]
    child2 = parent2[:split_point] + parent1[split_point:]

    return child1, child2


def mutation(chromosome):
    # Flip a random bit
    idx = random.randint(0, len(chromosome) - 1)
    if chromosome[idx] == 1:
        chromosome[idx] = 0
    else:
        chromosome[idx] = 1
    return chromosome


def main():
    # Initialize population
    population = []
    for i in range(population_size):
        population.append(generate_chromosome())

    # Evolution loop
    for generation in range(generations):
        print("\n--- Generation", generation + 1, "---")

        # Display current population
        for chromosome in population:
            fitness = calculate_fitness(chromosome)
            print(chromosome, "-> Fitness:", fitness)

        # Create new generation
        new_population = []

        while len(new_population) < population_size:
            # Select parents
            p1 = selection(population)
            p2 = selection(population)

            # Create offspring through crossover
            child1, child2 = crossover(p1, p2)

            # Apply mutation
            child1 = mutation(child1)
            child2 = mutation(child2)

            new_population.append(child1)
            if len(new_population) < population_size:  # Make sure we don't exceed population size
                new_population.append(child2)

        population = new_population

    # Find the best solution from final population
    best_solution = population[0]
    best_fitness_value = calculate_fitness(best_solution)

    for chrom in population:
        current_fitness = calculate_fitness(chrom)
        if current_fitness > best_fitness_value:
            best_solution = chrom
            best_fitness_value = current_fitness

    print("\n=== FINAL RESULTS ===")
    print("Best chromosome found:", best_solution)

    # Calculate final weight and value
    final_weight = 0
    final_value = 0
    for i in range(num_items):
        if best_solution[i] == 1:
            final_weight += weight_list[i]
            final_value += value_list[i]

    print("Total weight:", final_weight)
    print("Total value:", final_value)


# Run the algorithm
main()
```
