```python
#Parcels / Packages
packages=[1,2,3,4]
#cities
cities=['A','B','C']
#Capacity Of Each City
capacity={'A':2,'B':3,'C':1}
#Allowed Cities For Each Package
allowed_cities={1:['A','B'],2:['B','C'],3:['A','C'],4:['A']}
#Store Final Assignments
assignment={}
for i in range(len(cities)):
 city=cities[i]
 assignment[city]=[]

#Function For BackTracking
def package_assignment(index):
  #Return True if all package assigned
  if index==len(packages):
    return True

  pkg=packages[index]

  for j in range(len(allowed_cities[pkg])):
    var=allowed_cities[pkg][j]
    #Capacity Check_Point
    if len(assignment[var])<capacity[var]: #IF True Then Assign Package
     assignment[var].append(pkg)

     if package_assignment(index+1)==True:
      return True
     else:
      assignment[var].remove(pkg)
  return False


package_assignment(0)
print(assignment)
```
