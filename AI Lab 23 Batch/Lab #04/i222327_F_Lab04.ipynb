TASK# 01 ---> BFS Shortest Path

```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}


def bfs_shortest_path(graph, start, target):

    visited = set([start])
    queue = deque([(start, [start])])

    while queue:
        node, path = queue.popleft()

        if node == target:
            return path

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                new_path = path + [neighbor]
                queue.append((neighbor, new_path))

    return None


print(bfs_shortest_path(graph, 'A', 'F'))
```

```python
TASK# 02 ---> DFS Virus Spread
```

```python
def dfs_spread(graph, start):

    stack = [start]
    visited = set()
    infected = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)
            infected.append(node)

            for neighbor in graph[node][::-1]:  
                if neighbor not in visited:
                    stack.append(neighbor)

    return infected


graph = {
    'C1': ['C2', 'C3'],
    'C2': ['C1', 'C4'],
    'C3': ['C1'],
    'C4': ['C2', 'C5'],
    'C5': ['C4']
}

print(dfs_spread(graph, 'C1'))
```

```python
import heapq
import random

def uniform_cost_search(graph, start, goal):
    queue = []
    heapq.heappush(queue, (0, start, [start])) 

    visited = set()

    while queue:
        cost, node, path = heapq.heappop(queue)

        if node == goal:
            return path, cost

        if node in visited:
            continue

        visited.add(node)

        for neighbor, base_cost, construction in graph[node]:
            extra = random.randint(0, 3) if construction == "yes" else 0
            new_cost = cost + base_cost + extra
            new_path = path + [neighbor]

            heapq.heappush(queue, (new_cost, neighbor, new_path))

    return "Goal not reachable"


city_graph = {
    'S': [('A', 4, "no"), ('B', 7, "yes")],
    'A': [('C', 3, "no")],
    'B': [('C', 2, "yes"), ('D', 6, "no")],
    'C': [('E', 5, "yes")],
    'D': [('E', 4, "no")],
    'E': [('H', 6, "no")]
}

path, cost = uniform_cost_search(city_graph, 'S', 'H')
print("Path:", path)
print("Cost:", cost)
```
