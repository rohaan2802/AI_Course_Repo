The Tic-Tac-Toe game using the minimax algorithm

```python
import math

# Initialize an empty Tic-Tac-Toe board
board = [' ' for _ in range(9)]

# Function to display the board
def print_board():
    for row in [board[i*3:(i+1)*3] for i in range(3)]:
        print("| " + " | ".join(row) + " |")
    print()

# Function to check if there are empty spaces on the board
def is_empty_space():
    return ' ' in board

# Function to check for a winner
def check_winner(b, player):
    win_positions = [(0, 1, 2), (3, 4, 5), (6, 7, 8),
                     (0, 3, 6), (1, 4, 7), (2, 5, 8),
                     (0, 4, 8), (2, 4, 6)]
    return any(b[i] == b[j] == b[k] == player for i, j, k in win_positions)

# Function to evaluate the board
def evaluate():
    if check_winner(board, 'X'):
        return 1  # Maximizer (computer) wins
    elif check_winner(board, 'O'):
        return -1  # Minimizer (user) wins
    else:
        return 0  # Draw or undecided

# Minimax function
def minimax(is_maximizing):
    score = evaluate()

    # Terminal conditions
    if score == 1 or score == -1:  # Return score if game is won
        return score
    if not is_empty_space():  # Return 0 if it is a draw
        return 0

    if is_maximizing:  # Maximizing player's turn
        best_score = -math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                best_score = max(best_score, minimax(False))
                board[i] = ' '
        return best_score

    else:  # Minimizing player's turn
        best_score = math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                best_score = min(best_score, minimax(True))
                board[i] = ' '
        return best_score

# Function to find the best move for the computer
def best_move():
    best_score = -math.inf
    move = -1
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'X'
            score = minimax(False)
            board[i] = ' '
            if score > best_score:
                best_score = score
                move = i
    return move

# Main game loop
def play_game():
    print("Welcome to Tic Tac Toe! You are 'O', computer is 'X'")
    print_board()

    while is_empty_space():
        # User's move
        user_move = int(input("Enter your move (0-8): "))
        if board[user_move] == ' ':
            board[user_move] = 'O'
            print_board()
            if check_winner(board, 'O'):
                print("Congratulations! You won!")
                return

            # Computer's move
            computer_move = best_move()
            if computer_move != -1:
                board[computer_move] = 'X'
                print("Computer's move:")
                print_board()
                if check_winner(board, 'X'):
                    print("Computer wins!")
                    return

        if not is_empty_space():
            print("It's a draw!")
            return

# Start the game
play_game()
```
