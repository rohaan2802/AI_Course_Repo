```python
import random
import numpy as np
import math

# Constants
N = 10  # Grid size (N x N)
NUM_SCHOOLS = 3
NUM_HOSPITALS = 2
NUM_PARKS = 2
NUM_FIRE_STATIONS = 1

# Cooling parameters for Simulated Annealing
INITIAL_TEMPERATURE = 1000
COOLING_RATE = 0.95
MIN_TEMPERATURE = 1

# Grid values
EMPTY = 0
RESIDENTIAL = 1
RESTRICTED = -1
SCHOOL = 3
HOSPITAL = 4
PARK = 5
FIRE_STATION = 6


# Function to create an N x N grid
def create_grid(N):
    grid = np.zeros((N, N))

    # Randomly place residential areas (1/3 of the grid as residential)
    num_residential = (N * N) // 3
    residential_positions = random.sample([(i, j) for i in range(N) for j in range(N)], num_residential)
    for r in residential_positions:
        grid[r[0]][r[1]] = RESIDENTIAL

    # Randomly place restricted areas (1/5 of the grid as restricted)
    num_restricted = (N * N) // 5
    restricted_positions = random.sample([(i, j) for i in range(N) for j in range(N) if grid[i][j] == EMPTY], num_restricted)
    for r in restricted_positions:
        grid[r[0]][r[1]] = RESTRICTED

    return grid


# Function to print the grid in a readable format
def print_grid(grid):
    for row in grid:
        print(" ".join([str(int(cell)) for cell in row]))


# Function to calculate Manhattan distance between two points
def manhattan_distance(p1, p2):
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])


# Function to calculate the total cost (energy function)
def calculate_energy(grid, facility_positions, residential_positions):
    energy = 0
    for facility_type, positions in facility_positions.items():
        for pos in positions:
            # Sum distances to residential areas
            energy += sum(manhattan_distance(pos, res) for res in residential_positions)
            # Penalty for placing in restricted zones
            if grid[pos[0]][pos[1]] == RESTRICTED:
                energy += 100
    return energy


# Function to randomly place facilities on the grid
def place_facilities(grid, num_schools, num_hospitals, num_parks, num_fire_stations):
    facility_positions = {SCHOOL: [], HOSPITAL: [], PARK: [], FIRE_STATION: []}
    available_positions = [(i, j) for i in range(N) for j in range(N) if grid[i][j] == EMPTY]
    for _ in range(num_schools):
        pos = random.choice(available_positions)
        facility_positions[SCHOOL].append(pos)
        available_positions.remove(pos)
    for _ in range(num_hospitals):
        pos = random.choice(available_positions)
        facility_positions[HOSPITAL].append(pos)
        available_positions.remove(pos)
    for _ in range(num_parks):
        pos = random.choice(available_positions)
        facility_positions[PARK].append(pos)
        available_positions.remove(pos)
    for _ in range(num_fire_stations):
        pos = random.choice(available_positions)
        facility_positions[FIRE_STATION].append(pos)
        available_positions.remove(pos)
    return facility_positions


# Function to update the grid with facility positions
def update_grid_with_facilities(grid, facility_positions):
    for facility_type, positions in facility_positions.items():
        for pos in positions:
            grid[pos[0]][pos[1]] = facility_type
    return grid


# Function to perform simulated annealing
def simulated_annealing(grid, initial_facilities, residential_positions, initial_temp, cooling_rate, min_temp):
    current_positions = initial_facilities
    current_energy = calculate_energy(grid, current_positions, residential_positions)
    temperature = initial_temp

    while temperature > min_temp:
        new_positions = {k: list(v) for k, v in current_positions.items()}
        facility_type = random.choice(list(new_positions.keys()))
        if new_positions[facility_type]:
            idx = random.randint(0, len(new_positions[facility_type]) - 1)
            old_pos = new_positions[facility_type][idx]
            available_positions = [(i, j) for i in range(N) for j in range(N) if grid[i][j] == EMPTY and (i, j) != old_pos]
            if available_positions:
                new_pos = random.choice(available_positions)
                new_positions[facility_type][idx] = new_pos
                new_energy = calculate_energy(grid, new_positions, residential_positions)
                if new_energy < current_energy or random.random() < math.exp((current_energy - new_energy) / temperature):
                    current_positions = new_positions
                    current_energy = new_energy
        temperature *= cooling_rate
    return current_positions


# Dynamic zoning update simulation (adds new restrictions or changes population)
def dynamic_update(grid):
    # Randomly change residential areas to restricted or vice versa
    for _ in range(N // 2):
        i, j = random.randint(0, N - 1), random.randint(0, N - 1)
        if grid[i][j] == RESIDENTIAL:
            grid[i][j] = RESTRICTED
        elif grid[i][j] == EMPTY:
            grid[i][j] = RESIDENTIAL
    return grid


# Main function to run the optimization process
def run_simulation():
    # Step 1: Create the city grid
    grid = create_grid(N)
    print("Initial City Grid:")
    print_grid(grid)

    # Step 2: Place facilities
    facility_positions = place_facilities(grid, NUM_SCHOOLS, NUM_HOSPITALS, NUM_PARKS, NUM_FIRE_STATIONS)

    # Step 3: Identify residential positions
    residential_positions = [(i, j) for i in range(N) for j in range(N) if grid[i][j] == RESIDENTIAL]

    # Step 4: Run simulated annealing
    optimized_positions = simulated_annealing(grid, facility_positions, residential_positions, INITIAL_TEMPERATURE, COOLING_RATE, MIN_TEMPERATURE)

    # Step 5: Update the grid with final facility positions
    final_grid = update_grid_with_facilities(grid, optimized_positions)
    print("\nOptimized City Grid:")
    print_grid(final_grid)

    # Step 6: Simulate dynamic zoning update
    updated_grid = dynamic_update(final_grid)
    print("\nUpdated City Grid After Zoning Changes:")
    print_grid(updated_grid)


# Run the simulation
run_simulation()
```
